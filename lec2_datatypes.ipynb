```python
###python supports three types of data types
#basic type (integer, bolian,float, complex and string)
#container type (list, tuple, sets and dictionary)
#user defined type (class)
```

```python
#integers
print(5)
print(1e309)# python can handle large numbers too
```

```python
# float
print(7.9)
print(1.7e+309)
```

```python
#tuple
print((1,2,3,4,5,6))
```
